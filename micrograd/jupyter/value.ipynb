```python
from micrograd.value import ScalarValue

# inputs
x1 = ScalarValue(-2.0)
x2 = ScalarValue(0.0)

# weights
w1 = ScalarValue(3.0)
w2 = ScalarValue(1.0)

# bias 
b = ScalarValue(6.88943843995395804)

# net 
x1w1 = x1*w1 
x2w2 = x2*w2 

x1w1x2w2 = x1w1 * x2w2

n = x1w1x2w2 + b

# Activation (relu)
o = n.relu()

# backward
o.backward()
```

```python
def print_data():
    print(f"\nx1.grad, w1.grad: {x1.grad, w1.grad}")
    print(f"\nx2.grad, w2.grad: {x2.grad, w2.grad}")
    print(f"\nx1w1.grad, x2w2.grad, x1w1x2w2.grad: {x1w1.grad, x2w2.grad, x1w1x2w2.grad}")
    print(f"\nn.grad, o.grad: {n.grad, o.grad}")

print_data()
```

```python
o.backward()
print_data()
```
