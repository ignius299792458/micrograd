```python
from micrograd.value import ScalarValue

# inputs
x1 = ScalarValue(-2.0)
x2 = ScalarValue(0.0)

# weights
w1 = ScalarValue(3.0)
w2 = ScalarValue(1.0)

# bias 
b = ScalarValue(6.88943843995395804)

# net 
x1w1 = x1*w1 
x2w2 = x2*w2 

x1w1x2w2 = x1w1 + x2w2

n = x1w1x2w2 + b

# Activation (relu)
o = n.relu()

# backward
o.backward()

def print_data_scalar_value():
    print(f"\nx1.grad, w1.grad: {x1.grad, w1.grad}")
    print(f"\nx2.grad, w2.grad: {x2.grad, w2.grad}")
    print(f"\nx1w1.grad, x2w2.grad, x1w1x2w2.grad: {x1w1.grad, x2w2.grad, x1w1x2w2.grad}")
    print(f"\nn.grad, o.grad: {n.grad, o.grad}")
    print(o.data)

print_data_scalar_value()
```

```python
import torch

# inputs (tensors)
x1t = torch.Tensor([-2.0]).double(); x1t.requires_grad = True
x2t = torch.Tensor([0.0]).double(); x2t.requires_grad = True

# weights (tensors)
w1t = torch.Tensor([3.0]).double(); w1t.requires_grad = True
w2t = torch.Tensor([1.0]).double(); w2t.requires_grad = True

# bias  tensor
bt = torch.Tensor([6.88943843995395804]).double(); bt.requires_grad = True

# net (tensor)

nt = x1t*w1t + x2t*w2t + bt

# Activation (relu) tensor
ot = nt.relu()

# backward
ot.backward()


print(f"\nx1t.grad, w1t.grad: {x1t.grad.item(), w1t.grad.item()}")
print(f"\nx2t.grad, w2t.grad: {x2t.grad.item(), w2t.grad.item()}")
print(ot.data.item())
print(f"ot.data, o.data: {ot.data.item(), o.data}")


```
